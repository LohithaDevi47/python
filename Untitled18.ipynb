```python
import socket  
import random  
import time 
s = socket.socket() 
s.bind(("localhost", 8001))  
s.listen(5) 


c, adr = s.accept() 
print("From address", str(adr), "connection has been established")  


n = int(input("Enter number of frames: ")) 
N = int(input("Enter window size: ")) 

seq = 1 
frame = 1  
for i in range(N): 
    print('Frame sent ->', frame)  
    c.send(str(frame).encode()) 
    frame += 1 
    time.sleep(2) 

timer = 5  
while frame <= n: 
    t = random.randint(1, 7) 
    msg = c.recv(1).decode()  
    msg = int(msg)  
    print("Frame", msg)  

    if timer > t: 
        print("Acknowledgement received") 
        print('Frame sent ->', frame)  
        c.send(str(frame).encode()) 
        seq += 1 
        frame += 1  
        time.sleep(2) 
    else: 
        print("Acknowledgement not received") 
        print('Frame resent ->', msg) 
        c.send(str(msg).encode())  
        time.sleep(2)
s.close()
```
